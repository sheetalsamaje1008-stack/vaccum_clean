```python
# Vacuum Cleaner Agent

# Get initial state
room_A = input("Enter status of Room A (clean/dirty): ").lower()
room_B = input("Enter status of Room B (clean/dirty): ").lower()

location = input("Enter vacuum starting location (A/B): ").upper()

# Convert input to Boolean state
rooms = {
    "A": room_A,
    "B": room_B
}

print("\n--- Vacuum Cleaner Agent Started ---")

while True:

    # 1. Check current room
    print(f"\nVacuum is in Room {location}")
    print(f"Room {location} is {rooms[location].upper()}")

    # 2. If current room is dirty, clean it
    if rooms[location] == "dirty":

        print(f"Action: SUCK")
        rooms[location] = "clean"

        print(f"Room {location} is now CLEAN")

    # 3. If current room is clean, move to other room
    else:

        other_room = "B" if location == "A" else "A"

        print(f"Action: MOVE {location} -> {other_room}")

        location = other_room

    # 4. Check whether both rooms are clean
    if rooms["A"] == "clean" and rooms["B"] == "clean":

        print("\n--- GOAL ACHIEVED ---")
        print("Room A = CLEAN")
        print("Room B = CLEAN")
        print(f"Vacuum = Room {location}")

        break
```
